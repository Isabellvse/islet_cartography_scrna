```python
# Path and system utilities
import os                    # Operating system interface
import sys                   # System-specific parameters and functions
import glob                  # File pattern matching
from pathlib import Path     # Object-oriented filesystem paths
from pyhere import here      # Reproducible project paths
import gc

# Single-cell data handling
import anndata as ad            # Core data structure for single-cell data
import scanpy as sc
```

```python
anndata_dir = str(here('data/anndata/'))
```

#### Load

```python
# Adata object
adata = ad.read_h5ad(os.path.join(anndata_dir, "AH_combined.h5ad"))
```

#### Broad annotaiton

```python
# Define annotation
cell_type_broad = {
    "0": "alpha",
    "1": "gamma",
    "2": "alpha",
    "3": "alpha",
    "4": "alpha",
    "5": "alpha",
    "6": "alpha",
    "7": "alpha",
    "8": "delta",
    "9": "beta",
    "10": "beta",
    "11": "beta",
    "12": "cycling",
    "13": "beta",
    "14_0": "acinar",
    "14_1": "acinar",
    "14_2": "acinar",
    "15": "myeloid",
    "16": "schwann",
    "17": "ductal",
    "18": "ductal",
    "19": "epsilon",
    "20": "stellate",
    "21": "stellate",
    "22_0": "vascular",
    "22_1": "vascular",
    "23": "vascular",
    "24": "mast"
}

# add annotation
adata.obs["cell_type_broad"] = adata.obs["leiden_1.5_subcluster"].map(cell_type_broad).astype("category")

# define colors
colors_broad = {
    'acinar': '#c5b0d5',
    'alpha': '#1f77b4',
    'beta': '#ff0000',
    'cycling': '#aec7e8',
    'delta': '#895129',
    'ductal': '#FF46A2',
    'vascular': '#ae0000',
    'epsilon': '#BCF646',
    'gamma': '#FFBB78',
    'mast': '#0000FF',
    'myeloid': '#ee7733',
    'schwann': '#BCBD22',
    'stellate': '#64F844'
}

# ensure annotation is categorical
adata.obs["cell_type_broad"] = adata.obs["cell_type_broad"].astype("category")

# assign colors in the correct category order
adata.uns["cell_type_broad_colors"] = [
    colors_broad[c] for c in adata.obs["cell_type_broad"].cat.categories
]
```

#### Annotation

```python
# Define annotation
cell_type = {
    "0": "alpha",
    "1": "gamma",
    "2": "alpha",
    "3": "alpha",
    "4": "alpha",
    "5": "alpha",
    "6": "alpha",
    "7": "alpha",
    "8": "delta",
    "9": "beta",
    "10": "beta",
    "11": "beta",
    "12": "cycling",
    "13": "beta",
    "14_0": "acinar",
    "14_1": "acinar_reg_plus",
    "14_2": "acinar",
    "15": "myeloid",
    "16": "schwann",
    "17": "ductal_mucin",
    "18": "ductal",
    "19": "epsilon",
    "20": "stellate_quiescent",
    "21": "stellate_activated",
    "22_0": "pericyte",
    "22_1": "pericyte",
    "23": "endothelial",
    "24": "mast"
}

# add annotation
adata.obs["cell_type"] = adata.obs["leiden_1.5_subcluster"].map(cell_type).astype("category")

# define colors
colors = {
    'acinar': '#c5b0d5',
    'acinar_reg_plus': '#FFCD01',
    'alpha': '#1f77b4',
    'beta': '#ff0000',
    'cycling': '#aec7e8',
    'delta': '#895129',
    'ductal': '#FF46A2',
    'ductal_mucin': '#F7B6D2',
    'pericyte': '#ae0000',
    'endothelial': '#D844F8',
    'epsilon': '#BCF646',
    'gamma': '#FFBB78',
    'mast': '#0000FF',
    'myeloid': '#ee7733',
    'schwann': '#BCBD22',
    'stellate_activated': '#64F844',
    'stellate_quiescent': '#9467BD'
}

# ensure annotation is categorical
adata.obs["cell_type"] = adata.obs["cell_type"].astype("category")

# assign colors in the correct category order
adata.uns["cell_type_colors"] = [
    colors[c] for c in adata.obs["cell_type"].cat.categories
]
```

#### Add disease detmined by hba1c % 

```python
adata.obs['disease_hba1c'] = adata.obs['disease_harmonized']  # start as a copy

mask = (adata.obs['disease_harmonized'] == 'nd') & (adata.obs['hba_1_c_percent'] >= 5.7)
adata.obs.loc[mask, 'disease_hba1c'] = 'pre'
```

#### Save object

```python
adata.write(os.path.join(anndata_dir, 'AH_combined.h5ad')) 
```
